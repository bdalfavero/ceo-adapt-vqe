```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("xxz_parallel_times.csv", index_col=None)
print(df.head())
```

```python
fig, ax = plt.subplots()

colors = ['k', 'r', 'g']
for i, l in enumerate(np.unique(df["l"])):
    print(f"l = {l}")
    df_l = df[df["l"] == l]
    ax.plot(df_l["ncores"], df_l["avg_time"], '.', color=colors[i], label=f"{l:d}")
ax.legend()
ax.set_xlabel("Number of cores")
ax.set_ylabel("Time (ns)")
```
